```python
plt.figure(figsize=(7,5))
plt.hist(data_info_df["Підсумкова оцінка"], bins=10, edgecolor="red")
plt.title("Гістограма оцінок студентів")
plt.xlabel("Оцінка")
plt.ylabel("Кількість студентів")
plt.show()
```

```python
high = len(data_info_df[data_info_df["Підсумкова оцінка"] >= 90])
medium = len(data_info_df[(data_info_df["Підсумкова оцінка"] >= 75) & (data_info_df["Підсумкова оцінка"] < 90)])
low = len(data_info_df[data_info_df["Підсумкова оцінка"] < 75])

plt.figure(figsize=(6,6))
plt.pie(
    [high, medium, low],
    labels=["90-100", "75-89", "0-74"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Розподіл оцінок")
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.bar(data_info_df["Ім'я"], data_info_df["Підсумкова оцінка"])
plt.title("Оцінки студентів")
plt.xlabel("Студенти")
plt.ylabel("Оцінка")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(data_info_df["Години підготовки"], data_info_df["Підсумкова оцінка"])
plt.title("Залежність оцінки від годин підготовки")
plt.xlabel("Години підготовки")
plt.ylabel("Підсумкова оцінка")
plt.grid(True)
plt.show()
```
